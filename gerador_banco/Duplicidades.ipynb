```python
import pandas as pd

similares = pd.read_csv('similares.csv')
```

```python
similares = similares.sort_values(['similarity', 'reference'], ascending=[False, True])
```

```python
filtered = similares[similares['similarity'] > .80].copy()
```

```python
filtered['reference_index'] = filtered['reference'].map(lambda it: int(it.split('-')[0]))
filtered['similar_index'] = filtered['similar'].map(lambda it: int(it.split('-')[0]))

filtered = filtered[filtered['reference_index'] < filtered['similar_index']].copy()

filtered['reference_lyric'] = ''
filtered['similar_lyric'] = ''
```

```python
#filtered.pivot(index="similar_index", columns=[]).sort_index()
```

```python
import os
from pathlib import Path
from IPython.display import display

def read_music(index, name):
    path = Path('../dados/musicas') if index <= 612 else Path('../dados/musicas2')
    return (path / name).read_text()

for index in filtered.index:
    filtered.loc[index, 'reference_lyric'] = read_music(filtered.loc[index]['reference_index'], filtered.loc[index]['reference'])
    filtered.loc[index, 'similar_lyric'] = read_music(filtered.loc[index]['similar_index'], filtered.loc[index]['similar'])
    
pd.set_option('display.max_colwidth', int(1e10))
pd.set_option('display.expand_frame_repr', True)

filtered.style.set_properties(**{
    'text-align': 'left',
    'white-space': 'pre-wrap',
})
filtered.to_csv('duplicidades.csv')
```

```python
####################
# Drop duplicates

duplicacoes = pd.read_csv('./duplicidades-tratado.csv')
duplicacoes_identificadas = duplicacoes[duplicacoes['Duplicados'] == True][['reference_index', 'similar_index']]

indexes = [
    '../dados/livros/2-Cantando_os_Apelos_da_Eucaristia_-_Volume_III/indice.csv',
    '../dados/livros/1-Animados pela Fé/indice.csv',
    '../dados/todo/3-Louvemos ao Senhor 2018/index.csv',
    '../dados/todo/4-Músicas das Campanhas da Fraternidade/index.csv',
]

for index in indexes:
    content = pd.read_csv(index, index_col=0)

    duplicacoes_do_livro = set(duplicacoes_identificadas['similar_index'])
    musicas_duplicadas = set(content.index) & set(duplicacoes_identificadas['similar_index'])
    
    print(musicas_duplicadas)
    for id in duplicacoes_identificadas[duplicacoes_identificadas.similar_index.isin(musicas_duplicadas)].index:
        reference_index = duplicacoes_identificadas.loc[id]['reference_index']
        similar_index = duplicacoes_identificadas.loc[id]['similar_index']

        content = content.rename(index={similar_index: reference_index})

    content = content.reset_index().drop_duplicates().set_index('index')
    content.to_csv(index)

    print('Duplicações restantes')
    a = content.reset_index().drop_duplicates().groupby('indice').count()
    display(a[a['index']>1])


```
