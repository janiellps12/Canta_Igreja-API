```python
class Musica:
    def __init__(self, path):
        self.path = path
        self.conteudo = self.path.read_text()
    
    @property
    def rotulos(self):
        linha_rotulos = self._linha_rotulos()

        return [r.strip() for r in linha_rotulos.replace('Rótulos: ', '').split(';') if r.strip() != '']

    def _linha_rotulos(self):
        linha_rotulos = next((l for l in self.conteudo.splitlines() if l.startswith('Rótulos: ')), None)
    
        if linha_rotulos is None:
            raise ValueError(f'{self.path} - Arquivo sem linha de rótulos')

        return linha_rotulos
    
    @rotulos.setter
    def rotulos(self, novos_rotulos):
        nova_linha_rotulos = 'Rótulos: ' + '; '.join(novos_rotulos)

        self.conteudo = self.conteudo.replace(self._linha_rotulos(), nova_linha_rotulos)
        self.path.write_text(self.conteudo)

    def validar_rotulos(self):
        try:
            Rotulos.validar_rotulos(self.rotulos)
        except ValueError as e:
            raise ValueError(f"Música {self.id_musica}: {e}")
```

```python
from pathlib import Path
import pandas as pd

musicas = []

for file in Path('../../dados/musicas').glob('*.txt'):
    musica = Musica(file)
    rotulos = musica.rotulos

    erros = []
    try:
        musica.validar_rotulos()
    except Exception as e:
        erros.append(str(e))

    musicas.append([musica.path.name, rotulos, erros])

musicas = pd.DataFrame(musicas, columns=['arquivo', 'rotulos', 'erros'])
```

```python
def atualizar_arquivo(serie, indice_arquivo='indice_geral', colunas=['Momento', 'Calendário', 'Outros']):
    rotulos = []

    for coluna in colunas:
        rotulos += serie[coluna].split(';') if pd.notna(serie[coluna]) else []

    rotulos = [r.strip() for r in rotulos]

    if rotulos == []:
        return

    if len(set(rotulos) - Rotulos.rotulos_possiveis) > 0:
        raise ValueError(f'{serie[indice_arquivo]} - Rótulos inválidos: {set(rotulos) - Rotulos.rotulos_possiveis}')

    try:
        musica = Musica(Path(f'../../dados/musicas/{serie[indice_arquivo]}.txt'))
    except Exception as e:
        print(f'Erro ao abrir arquivo {serie[indice_arquivo]}: {e}')
        return
   
    rotulos_cadastrados = musica.rotulos
    
    rotulos_finais = sorted(set(rotulos_cadastrados) | set(rotulos))

    if len(set(rotulos_finais) - Rotulos.rotulos_possiveis) > 0:
        raise ValueError(f'{serie[indice_arquivo]} - Rótulos inválidos: {set(rotulos_finais) - Rotulos.rotulos_possiveis} - Algum rótulo já cadastrado é inválido.')

    musica.rotulos = rotulos_finais
    
#atualizar_arquivo(data.iloc[0])

data.apply(atualizar_arquivo, axis=1)
```
